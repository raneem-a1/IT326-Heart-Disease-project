## 4. Outliers and Boxplots

```python
# Create boxplots for numerical attributes to identify potential outliers

import matplotlib.pyplot as plt
import seaborn as sns

numerical_attributes = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']

plt.figure(figsize=(12, 8))

for i, column in enumerate(numerical_attributes, 1):
    plt.subplot(2, 3, i)
    sns.boxplot(y=df[column])
    plt.title(f'Boxplot of {column}')
    plt.ylabel(column)

plt.tight_layout()
plt.show()
```

```python
# Detect and count outliers using the IQR method

outlier_summary = []

for column in numerical_attributes:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    outlier_summary.append({
        'Attribute': column,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Number of Outliers': len(outliers)
    })

outlier_summary_df = pd.DataFrame(outlier_summary)
outlier_summary_df
```

### Outliers Analysis

The boxplots and IQR method were used to identify potential outliers in the numerical attributes.

- **age:** No outliers were detected.
- **trestbps:** 30 outliers were detected, indicating the presence of unusually high resting blood pressure values.
- **chol:** 16 outliers were detected, representing unusually high cholesterol values.
- **thalach:** 4 outliers were detected, representing unusually low maximum heart rate values.
- **oldpeak:** 7 outliers were detected, mainly at the higher end of the distribution.

Overall, most numerical attributes contain a relatively small number of potential outliers, while **trestbps** has the highest number. These outliers should be considered during the preprocessing stage because extreme values may affect the analysis and the performance of some data mining techniques.

# Q2. Data Preprocessing

## 1 . Variable Transformation

Log Transformation was applied to a numerical attribute with a skewed distribution.

**Why?**  
Log transformation helps reduce the effect of extreme values and reduce skewness in the distribution.

**Which attribute?**  
The `oldpeak` attribute was selected because its distribution is right-skewed, with most values concentrated near zero and fewer higher values.

**How?**  
Log transformation was applied to `oldpeak` using NumPy's `log1p()` function.



```python
# Apply Log Transformation on oldpeak


df_transformed = df.copy()

df_transformed['oldpeak'] = np.log1p(df_transformed['oldpeak'])

df_transformed.head()
```

```python
# Plot oldpeak after Log Transformation

df_transformed['oldpeak'].hist(figsize=(10, 5))

plt.xlabel('oldpeak')
plt.ylabel('Frequency')
plt.title('oldpeak After Log Transformation')

plt.show()
```

###Variable Transformation
**Result:**  
The transformation reduces the influence of higher values and makes the distribution less skewed while preserving the original information.



## 2. Normalization

Normalization is a preprocessing technique used to scale numerical attributes into a common range.

### Why?
The dataset contains numerical attributes with different ranges, such as age, cholesterol, and maximum heart rate. Normalization helps prevent attributes with larger values from dominating the analysis.

### Which attributes?
Normalization is applied to the following numerical attributes:
- age
- trestbps
- chol
- thalach
- oldpeak

### How?
Min-Max normalization is applied using MinMaxScaler.



```python
from sklearn.preprocessing import MinMaxScaler

# Create a copy of the transformed dataset
df_normalized = df_transformed.copy()

# Select numerical attributes for normalization
numerical_columns = [
    'age',
    'trestbps',
    'chol',
    'thalach',
    'oldpeak'
]

# Apply Min-Max Normalization
scaler = MinMaxScaler()

df_normalized[numerical_columns] = scaler.fit_transform(
    df_normalized[numerical_columns]
)

df_normalized.head()
```

###Normalization
**Result:**

The numerical attributes were scaled to a common range, making them more suitable for further analysis and machine learning tasks.

```python
# Save the Log_Normalized dataset

df_normalized.to_csv("Log_Normalized_dataset.csv", index=False)

print("Log_Normalized dataset saved successfully!")
```

## 3.Additional Preprocessing Technique

### Discretization

Discretization converts numerical values into categories/ranges.

It helps simplify the data and improve interpretation by grouping continuous numerical values into meaningful categories.

**Why?**  
Discretization was applied to simplify the age values and make them easier to interpret as age groups instead of continuous numerical values.

**Which attribute?**  
The `age` attribute was selected because it is a continuous numerical attribute.

**How?**  
The normalized `age` values, which range from 0 to 1, were divided into three ranges using `pd.cut()`:

- **Young:** 0 to 0.4
- **Middle-aged:** 0.4 to 0.7
- **Older:** 0.7 to 1





```python
# Apply Discretization on age

df_final = df_normalized.copy()

df_final['age_group'] = pd.cut(
    df_final['age'],
    bins=[0, 0.4, 0.7, 1],
    labels=['Young', 'Middle-aged', 'Older'],
    include_lowest=True
)

df_final.head()
```

###Discretization
**Result:**  

A new categorical attribute called `age_group` was created with three categories: Young, Middle-aged, and Older.

```python
# Check the distribution of age groups
df_final['age_group'].value_counts(dropna=False)
```

```python
# Save Preprocessed dataset

df_final.to_csv("Preprocessed_dataset.csv", index=False)

print("Preprocessed dataset saved successfully!")
```
