## Project Goal

The goal of this project is to apply classification and clustering techniques to a heart disease dataset.

For the classification task, the objective is to predict whether a patient has heart disease based on clinical and demographic attributes. The target variable contains two classes: 0 represents the absence of heart disease, while 1 represents the presence of heart disease.

For the clustering task, the objective is to group patients based on similarities in their health characteristics and investigate the patterns and characteristics of the resulting clusters.

## Dataset Source

The Heart Disease dataset used in this project was sourced from Kaggle.

**Source:** https://www.kaggle.com/datasets/johnsmith88/heart-disease-dataset
